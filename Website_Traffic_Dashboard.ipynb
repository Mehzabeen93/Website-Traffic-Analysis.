```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("../Dataset/Website_Traffic_Analysis_Dataset.csv")

df.head()
```

```python
print("Total Sessions:", len(df))
print("Total Users:", df["User_ID"].nunique())
print("Total Conversions:", df["Conversion"].sum())
```

```python
total_sessions = len(df)

total_users = df["User_ID"].nunique()

total_conversions = df["Conversion"].sum()

bounce_rate = df["Bounce"].mean() * 100

conversion_rate = (total_conversions / total_sessions) * 100

avg_duration = df["Session_Duration"].mean()

avg_pages = df["Pages_Per_Session"].mean()

print("Total Sessions:", total_sessions)
print("Total Users:", total_users)
print("Total Conversions:", total_conversions)
print("Bounce Rate:", round(bounce_rate, 2), "%")
print("Conversion Rate:", round(conversion_rate, 2), "%")
print("Average Session Duration:", round(avg_duration, 2), "seconds")
print("Average Pages Per Session:", round(avg_pages, 2))
```

```python
fig = px.scatter(
    x=[0, 1, 2, 3],
    y=[1, 1, 1, 1],
    size=[1, 1, 1, 1],
    color=["Sessions", "Users", "Bounce Rate", "Conversion Rate"],
    text=[
        f"Sessions<br>{total_sessions:,}",
        f"Users<br>{total_users:,}",
        f"Bounce Rate<br>{bounce_rate:.2f}%",
        f"Conversion Rate<br>{conversion_rate:.2f}%"
    ],
    size_max=80
)

fig.update_traces(
    mode="markers+text",
    textposition="middle center"
)

fig.update_layout(
    title="Website Traffic KPIs",
    showlegend=False,
    xaxis_visible=False,
    yaxis_visible=False,
    height=350
)

fig.show()
```

```python
# Calculate sessions by traffic source

traffic_source = df["Traffic_Source"].value_counts()

print(traffic_source)
```

```python
fig = px.bar(
    x=traffic_source.index,
    y=traffic_source.values,
    title="Website Traffic by Source",
    labels={
        "x": "Traffic Source",
        "y": "Total Sessions"
    },
    text=traffic_source.values
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
source_conversion = df.groupby("Traffic_Source").agg(
    Total_Sessions=("Session_ID", "count"),
    Total_Conversions=("Conversion", "sum")
)

source_conversion["Conversion_Rate"] = (
    source_conversion["Total_Conversions"] /
    source_conversion["Total_Sessions"] * 100
)

source_conversion.round(2)
```

```python
fig = px.bar(
    source_conversion.reset_index(),
    x="Traffic_Source",
    y="Conversion_Rate",
    title="Conversion Rate by Traffic Source",
    text_auto=".2f",
    labels={
        "Traffic_Source": "Traffic Source",
        "Conversion_Rate": "Conversion Rate (%)"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
device_summary = df.groupby("Device").agg(
    Total_Sessions=("Session_ID", "count"),
    Total_Conversions=("Conversion", "sum"),
    Average_Duration=("Session_Duration", "mean")
)

device_summary.round(2)
```

```python
fig = px.pie(
    device_summary.reset_index(),
    names="Device",
    values="Total_Sessions",
    title="Website Traffic by Device",
    hole=0.4
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
device_summary["Conversion_Rate"] = (
    device_summary["Total_Conversions"] /
    device_summary["Total_Sessions"] * 100
)

device_summary.round(2)
```

```python
fig = px.bar(
    device_summary.reset_index(),
    x="Device",
    y="Conversion_Rate",
    title="Conversion Rate by Device",
    text_auto=".2f",
    labels={
        "Conversion_Rate": "Conversion Rate (%)"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Date"].dtype
```

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)

df.head()
```

```python
monthly_traffic = df.groupby("Month")["Session_ID"].count()

monthly_traffic
```

```python
fig = px.line(
    x=monthly_traffic.index,
    y=monthly_traffic.values,
    title="Monthly Website Traffic Trend",
    markers=True,
    labels={
        "x": "Month",
        "y": "Total Sessions"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
monthly_conversions = df.groupby("Month")["Conversion"].sum()

monthly_conversions
```

```python
fig = px.bar(
    x=monthly_conversions.index,
    y=monthly_conversions.values,
    title="Monthly Website Conversions",
    labels={
        "x": "Month",
        "y": "Total Conversions"
    },
    text=monthly_conversions.values
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
monthly_summary = df.groupby("Month").agg(
    Total_Sessions=("Session_ID", "count"),
    Total_Conversions=("Conversion", "sum")
)

monthly_summary["Conversion_Rate"] = (
    monthly_summary["Total_Conversions"] /
    monthly_summary["Total_Sessions"] * 100
)

monthly_summary.round(2)
```

```python
fig = px.line(
    monthly_summary.reset_index(),
    x="Month",
    y="Conversion_Rate",
    title="Monthly Conversion Rate Trend",
    markers=True,
    text=monthly_summary["Conversion_Rate"].round(2),
    labels={
        "Month": "Month",
        "Conversion_Rate": "Conversion Rate (%)"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
page_summary = df.groupby("Page_Visited").agg(
    Total_Sessions=("Session_ID", "count"),
    Total_Conversions=("Conversion", "sum"),
    Average_Duration=("Session_Duration", "mean"),
    Bounce_Rate=("Bounce", "mean")
)

page_summary["Bounce_Rate"] = page_summary["Bounce_Rate"] * 100

page_summary["Conversion_Rate"] = (
    page_summary["Total_Conversions"] /
    page_summary["Total_Sessions"] * 100
)

page_summary = page_summary.round(2)

page_summary
```

```python
fig = px.bar(
    page_summary.reset_index(),
    x="Page_Visited",
    y="Conversion_Rate",
    title="Conversion Rate by Website Page",
    text_auto=".2f",
    labels={
        "Page_Visited": "Website Page",
        "Conversion_Rate": "Conversion Rate (%)"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
engagement_summary = df.groupby("Page_Visited").agg(
    Average_Duration=("Session_Duration", "mean"),
    Average_Pages=("Pages_Per_Session", "mean")
)

engagement_summary.round(2)
```

```python
fig = px.bar(
    engagement_summary.reset_index(),
    x="Page_Visited",
    y="Average_Duration",
    title="Average Session Duration by Website Page",
    text_auto=".2f",
    labels={
        "Page_Visited": "Website Page",
        "Average_Duration": "Average Duration (Seconds)"
    }
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```
